## 1. NumPy Array Creation and Dimensions

NumPy supports multidimensional arrays. The `ndim` attribute
returns the number of dimensions, while `shape` describes
the size along each dimension.

```python
# Create a 1D array
one_dimensional = np.array([10, 20, 30, 40])

# Create a 2D array
two_dimensional = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# Create a 3D array
three_dimensional = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

# Inspect array properties
arrays = {
    "1D": one_dimensional,
    "2D": two_dimensional,
    "3D": three_dimensional
}

for name, array in arrays.items():
    print(f"\n{name} Array:")
    print(array)
    print("Shape:", array.shape)
    print("Dimensions:", array.ndim)
    print("Data type:", array.dtype)
```

## 2. Broadcasting, Vectorisation and Matrix Multiplication

### Broadcasting
Broadcasting allows NumPy to perform operations on arrays
with compatible shapes without explicitly repeating data.

### Vectorisation
Vectorisation applies operations to entire arrays without
writing explicit Python loops.

### Matrix Multiplication
Matrix multiplication combines rows of the first matrix
with columns of the second matrix. The inner dimensions
must match.

```python
# Broadcasting
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

offset = np.array([1, 2, 3])
broadcast_result = matrix + offset

print("Broadcasting result:")
print(broadcast_result)


# Vectorisation
numbers = np.array([1, 2, 3, 4, 5])

squared = numbers**2
scaled = numbers * 10

print("\nOriginal array:", numbers)
print("Squared:", squared)
print("Scaled:", scaled)


# Matrix multiplication
first_matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

second_matrix = np.array([
    [7, 8],
    [9, 10],
    [11, 12]
])

product = first_matrix @ second_matrix

print("\nMatrix multiplication result:")
print(product)
print("Product shape:", product.shape)
```

## 3. Statistical Analysis Using the Iris Dataset

The Iris dataset contains 150 observations and four numerical
features.

We use NumPy to calculate:

- Mean: Average value of each feature.
- Standard deviation: Measures the spread of values around the mean.
- Correlation: Measures the strength and direction of linear
  relationships between features.

```python
# Locate the Iris dataset
data_path = Path("../data/iris.csv")

feature_names = [
    "Sepal length",
    "Sepal width",
    "Petal length",
    "Petal width"
]

# Load the four numerical columns
measurements = np.genfromtxt(
    data_path,
    delimiter=",",
    usecols=(0, 1, 2, 3),
    dtype=np.float64,
    autostrip=True
)

measurements = np.atleast_2d(measurements)

# Validate the dataset
if not np.isfinite(measurements).all():
    raise ValueError("Dataset contains invalid measurements.")

# Calculate statistics
means = np.mean(measurements, axis=0)
standard_deviations = np.std(measurements, axis=0)
correlation_matrix = np.corrcoef(measurements, rowvar=False)

print("Dataset shape:", measurements.shape)

print("\nMean:")
for name, value in zip(feature_names, means):
    print(f"{name}: {value:.3f}")

print("\nStandard deviation:")
for name, value in zip(feature_names, standard_deviations):
    print(f"{name}: {value:.3f}")

print("\nCorrelation matrix:")
print(np.round(correlation_matrix, 3))
```

### Q1. What is broadcasting in NumPy?

Broadcasting allows NumPy to perform arithmetic operations
between arrays of compatible shapes without explicitly
duplicating their data.

Example:
A matrix with shape (2, 3) can be added to a vector
with shape (3,). The vector is applied to each row.

### Q2. Why is vectorised NumPy faster than Python loops?

NumPy performs operations using optimized, compiled numerical
routines and processes array elements efficiently in bulk.
This reduces Python-level loop overhead.

### Q3. What is the difference between a NumPy array and
a Pandas Series?

A NumPy array is primarily an N-dimensional numerical structure
with efficient mathematical operations.

A Pandas Series is a one-dimensional labelled data structure
with an index, making it useful for working with tabular data.
